```python
from sqlalchemy import create_engine
import urllib
```

```python
import pyodbc

# Filters for SQL Server drivers installed on your system
sql_drivers = [d for d in pyodbc.drivers() if 'SQL Server' in d]
print(sql_drivers)
```

```python
server = 'OMKAR\SQLEXPRESS'
database = 'customer_behaviour'
driver = 'ODBC Driver 17 for SQL Server'

params = urllib.parse.quote_plus(
    f"DRIVER={{{driver}}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
    f"TrustServerCertificate=yes;" # Prevents SSL certificate errors in ODBC 18
)
```

```python
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}", fast_executemany=True)
```

```python
table_name = 'customer'

df.to_sql(
    name=table_name,
    con=engine,
    schema='dbo',
    if_exists='replace',  # Options: 'fail', 'replace' (drops & recreates), 'append'
    index=False,          # Prevents writing the pandas DataFrame index as a column
)
print(f"Successfully loaded {len(df)} rows into '{table_name}'.")
```

```python
pd.read_sql("SELECT TOP 5 * FROM customer;", engine)
```
